```python
import pandas as pd
import numpy as np


import matplotlib.pyplot as plt
plt.style.use('ggplot')
%matplotlib inline

from sklearn import linear_model
from sklearn.ensemble import RandomForestRegressor

from sklearn import model_selection
from sklearn.tree import DecisionTreeRegressor, DecisionTreeClassifier, export_graphviz
from sklearn import metrics
from sklearn.feature_selection import SelectFromModel, RFECV
import urllib.request

def get_regression_metrics(model, actual, predicted):

    """Calculate main regression metrics.

    Args:
        model (str): The model name identifier
        actual (series): Contains the test label values
        predicted (series): Contains the predicted values

    Returns:
        dataframe: The combined metrics in single dataframe


    """
    regr_metrics = {
                        'Root Mean Squared Error' : metrics.mean_squared_error(actual, predicted)**0.5,
                        'Mean Absolute Error' : metrics.mean_absolute_error(actual, predicted),
                        'R^2' : metrics.r2_score(actual, predicted),
                        'Explained Variance' : metrics.explained_variance_score(actual, predicted)
                   }

    #return reg_metrics
    df_regr_metrics = pd.DataFrame.from_dict(regr_metrics, orient='index')
    df_regr_metrics.columns = [model]
    return df_regr_metrics

urllib.request.urlretrieve("https://ethaneldridge.github.io/cisc499/Module11-Jupyter%20Notebooks%20(Part%201%20of%203)/resources/Module11_Dataset_Maintenance_Train.csv", './train.data')
df_train = pd.read_csv('./train.data')

urllib.request.urlretrieve("https://ethaneldridge.github.io/cisc499/Module11-Jupyter%20Notebooks%20(Part%201%20of%203)/resources/Module11_Dataset_Maintenance_Test.csv", './test.data')
df_test = pd.read_csv('./test.data')

#Let us prepare data for regression model

#These are original features
features_orig = ['setting1','setting2','setting3','s1','s2','s3','s4','s5','s6','s7','s8','s9','s10','s11','s12','s13','s14','s15','s16','s17','s18','s19','s20','s21']

#These are features with low or no correlation with regression label
features_lowcr = ['setting3', 's1', 's10', 's18','s19','s16','s5', 'setting1', 'setting2']

#These are features that have correlation with regression label
features_corrl = ['s2', 's3', 's4', 's6', 's7', 's8', 's9', 's11', 's12', 's13', 's14', 's15', 's17', 's20','s21']

#features is a variable to hold the set of features to experiment with
features = features_orig

X_train = df_train[features]
y_train = df_train['ttf']

X_test = df_test[features]
y_test = df_test['ttf']

print('Task #03 Completed!')
```

```python
import urllib.request
import pandas as pd

# 1. Download dataset
urllib.request.urlretrieve("http://archive.ics.uci.edu/ml/machine-learning-databases/iris/iris.data", './iris.data')

# 2. Open csv file and store in DataFrame named df (no header column)
df = pd.read_csv('./iris.data', header=None)

# 3. Print first five rows to check dataset
print("First 5 rows of the dataset:")
print(df.head())

# 4. Add column names
df.columns = ['sepal_length', 'sepal_width', 'petal_length', 'petal_width', 'class']

print("\nDataFrame with column names:")
print(df.head())
```

```python
# load a subset of hard-coded data as fallback to Task #12
data = [
        [5.1,	3.5,	1.4,	0.2,	'Iris-setosa'],
        [4.9,	3,	1.4,0.2,	'Iris-setosa'],
        [4.7,	3.2,	1.3,	0.2,	'Iris-setosa'],
        [4.6,	3.1,	1.5,	0.2,	'Iris-setosa'],
        [5,	3.6,	1.4,	0.2,	'Iris-setosa'],
        [7,	3.2,	4.7,	1.4,	'Iris-versicolor'],
        [6.4,	3.2,	4.5,	1.5,	'Iris-versicolor'],
        [6.9,	3.1,	4.9,	1.5,	'Iris-versicolor'],
        [5.5,	2.3,	4,	1.3,	'Iris-versicolor'],
        [6.5,	2.8,	4.6,	1.5,	'Iris-versicolor'],
        [6.3,	3.3,	6,	2.5,	'Iris-virginica'],
        [5.8,	2.7,	5.1,	1.9,	'Iris-virginica'],
        [7.1,	3,	5.9,	2.1,	'Iris-virginica'],
        [6.3,	2.9,	5.6,	1.8,	'Iris-virginica'],
        [6.5,	3,	5.8,	2.2,	'Iris-virginica']
]

# Create the pandas DataFrame
df = pd.DataFrame(data, columns = ['sepal_length', 'sepal_width', 'petal_length', 'petal_width', 'class'])

# print dataframe.
df
```

Task #13 (0.5 Pts): Execute the following code to train the KNN Classifier and make a prediction

```python
from sklearn.neighbors import KNeighborsClassifier

# Dictionary to input the different numbers for different classes
label_encode = {"class": {"Iris-setosa":1, "Iris-versicolor":2, "Iris-virginica":3}}

# Use .replace to change the different classes into numbers
df.replace(label_encode,inplace=True)

# Initialise the KNeighborsClassifier
KNN = KNeighborsClassifier()

# Extract out the x values and y values. x will be sepal_length and y will be classes
x = df[['sepal_length','sepal_width']]
y = df['class']

# Print .head() of x and y to make sure the data is correct
print(x.head())
print(y.head())

# Train KNN using the x and y values. This is done through the .fit method.
KNN = KNN.fit(x,y)

# Let us use the trained KNN to predict the type of flower if its sepal length = 5 and sepal_width = 3 We can use the .predict method to do so.
test = pd.DataFrame()
test['sepal_length'] = [5]
test['sepal_width'] = [3]
predict_flower = KNN.predict(test)

# Print predict_flower
print(predict_flower)
```
